## **Section 1. Business Understanding**

### **1.1 Context**

Alpha Company adalah perusahaan e-commerce berskala menengah yang tengah menghadapi tantangan kritis berupa **tingginya tingkat churn pelanggan**, yaitu kondisi ketika pelanggan berhenti bertransaksi atau berpindah ke kompetitor.
[Churn pelanggan merupakan ancaman besar terhadap profitabilitas bisnis](https://www.shipscience.com/understanding-customer-churn-rate-how-to-measure-it-and-why-it-matters-for-e-commerce-operations/), karena tidak hanya mengakibatkan kehilangan pendapatan langsung, tetapi juga menimbulkan **biaya tambahan yang signifikan untuk memperoleh pelanggan baru**.
[Menurut studi industri]((https://www.opensend.com/post/customer-acquisition-cost-ecommerce)), **biaya akuisisi pelanggan (Customer Acquisition Cost / CAC)** di sektor e-commerce berkisar antara **\$50–\$130 per pelanggan**, sedangkan **biaya retensi pelanggan (Customer Retention Cost / CRC)** rata-rata **[_3–6 kali lebih rendah_](https://www.totango.com/blog/how-to-calculate-the-cost-of-customer-retention-versus-customer-acquisition?)** daripada akuisisi.

[Churn rate adalah salah satu metrik yang menunjukkan arah keberlanjutan suatu perusahaan](https://wesupplylabs.com/the-ultimate-guide-to-customer-churn-in-ecommerce/). Oleh karena itu, churn rate bisa menjadi indikasi bahwa perusahaan sudah memberikan layanan / produk yang bagus di masyarakat. Nilai churn rate sendiri dapat berbeda - beda untuk masing - masing jenis bisnis. Pada umumnya, nilai churn rate bervariasi dan [batasan 5% - 10%](https://www.yotpo.com/blog/churn-rate/#:~:text=General%20eCommerce%20Benchmarks,be%20looked%20into%20right%20away) biasanya dapat diterima untuk jenis tipe perusahaan ecommerce. Berdasarkan [suatu studi](https://intl.delight.fit/blogs/insight/customer-retention-through-proactive-care-why-a-5-increase-in-retention-can-boost-profits-by-95?shpxid=5f51ce91-28fd-46fe-9928-b8c3282628b2), dengan mengurangi nilai churn rate sebesar *5%* dapat meningkatkan keuntungan sampai dengan *25% - 95%*.

Sebagai perusahaan e-commerce menengah, Alpha Company diperkirakan memiliki **CAC sekitar \$80 per pelanggan baru**, sementara **biaya retensi (CRC)** hanya sekitar **\$20 per pelanggan per tahun**.
Perbandingan ini menunjukkan bahwa **menjaga pelanggan yang sudah ada jauh lebih efisien** dibandingkan menarik pelanggan baru, yang menjadikan retensi pelanggan sebagai pilar utama keberlanjutan bisnis.

Mengurangi churn bukan sekadar meningkatkan kepuasan pelanggan, tetapi juga strategi untuk memperkuat **Customer Lifetime Value (CLV)**, yaitu total nilai ekonomi yang dihasilkan pelanggan sepanjang hubungan mereka dengan perusahaan.
[Pemahaman yang lebih mendalam tentang churn dan CLV membantu perusahaan memusatkan upaya pada pelanggan bernilai tinggi, merancang strategi retensi yang tepat sasaran, dan memaksimalkan margin keuntungan jangka panjang](https://seller.alibaba.com/id/businessblogs/what-is-customer-lifetime-value-clv-the-key-to-success-in-ecommerce-px002c29x).

Melalui pendekatan berbasis data, Alpha Company berupaya **mengidentifikasi pelanggan berisiko churn sejak dini**, memahami faktor-faktor perilaku dan demografis yang memengaruhinya, serta mengembangkan intervensi yang dapat meningkatkan keterlibatan pelanggan (*customer engagement*), memperkuat loyalitas, dan menurunkan churn secara berkelanjutan.

### **1.6 Metric Evaluation (Business Metric, Machine Learning Evaluation Metric)**

Dalam konteks churn prediction, dua jenis kesalahan utama yang dapat terjadi adalah:

**Type 1 Error / False Positive (FP):**  
- Model memprediksi pelanggan akan churn, padahal sebenarnya tidak.  
- **Konsekuensi bisnis:** Perusahaan mungkin mengalokasikan sumber daya retensi (voucher, promo, outreach) kepada pelanggan yang sebenarnya masih loyal → mengurangi efisiensi biaya.

**Type 2 Error / False Negative (FN):**  
- Model memprediksi pelanggan tidak akan churn, padahal sebenarnya churn.  
- **Konsekuensi bisnis:** Perusahaan kehilangan pelanggan secara nyata tanpa tindakan pencegahan → menyebabkan potensi kehilangan pendapatan dan peningkatan biaya akuisisi pelanggan baru.

**Metrik Evaluasi: F2-Score**

Karena perusahaan berfokus untuk **menangkap sebanyak mungkin pelanggan yang benar-benar berisiko churn (minimasi FN)** namun **tetap mempertahankan efisiensi biaya retensi (kontrol FP)**, metrik yang paling sesuai adalah **F2-Score**.

**F2-Score** memberikan bobot **lebih besar pada Recall** (mendeteksi sebanyak mungkin pelanggan churn) dibanding Precision, namun tetap mempertahankan keseimbangan antara keduanya.

Uji normalitas akan dilakukan menggunakan **Saphiro-Wilk** karena mengingat ukuran dataset yang terdiri dari sekitar 5rb baris. Selain itu **Saphiro-Wilk** juga lebih *robust* terhadap data berdistribusi *skewed*. Hal tersebut sesuai jika melihat pada histogram diatas, fitur numerik dataset yang kita miliki didominasi oleh data dengan distribusi *right-skewed*.

```python
results = []

for col in num_cols:
    x = df[col].dropna().values

    if len(x) >= 3:  # Shapiro-Wilk butuh minimal 3 observasi
        stat, p = stats.shapiro(x)
        skew = stats.skew(x, nan_policy='omit', bias=False)
    else:
        p = np.nan
        skew = np.nan

    results.append({
        'feature': col,
        'n': len(x),
        'pValue': p,
        'skewness': skew
    })

shapiro_df = (
    pd.DataFrame(results)
      .set_index('feature')
      .assign(verdict=lambda d: np.where(d['pValue'] > 0.05, 'Normal', 'Not Normal'))
      .sort_values('pValue')
)

display(
    shapiro_df
      .round({'pValue': 5, 'skewness': 3})
)
```

```python
iqr_results = {}

for col in num_cols:
    data = df[col].dropna()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[(data < lower_bound) | (data > upper_bound)]

    iqr_results[col] = {
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'lowerBound': lower_bound,
        'upperBound': upper_bound,
        'outlierCount': outliers.count(),
        'outlierPercentage': round((outliers.count() / len(data)) * 100, 2)
    }

iqr_df = pd.DataFrame(iqr_results).T.sort_values('outlierPercentage', ascending=False)
display(iqr_df)
```

Secara menyeluruh, fitur numerikal pada dataset ini memiliki distribusi tidak normal. Kemudian, untuk mendeteksi keberadaan nilai ekstrem pada fitur numerik, dilakukan analisis menggunakan metode **Interquartile Range (IQR)**. Metode ini menghitung rentang antara kuartil ke-1 (Q1) dan kuartil ke-3 (Q3) untuk menentukan batas bawah dan batas atas distribusi yang dianggap wajar.

Dari hasil analisis IQR, beberapa fitur numerik menunjukkan adanya anomali dengan proporsi yang bervariasi. Berikut ringkasan hasilnya:

* **Fitur dengan outlier tertinggi:**

  * `Churn` memiliki **16.84%** data yang terdeteksi sebagai outlier. Namun karena `Churn` adalah variabel kategorikal biner (0/1), nilai ekstrem disini bukanlah *anomali sebenarnya* melainkan efek dari *skala biner*, **sehingga tidak perlu diperlakukan sebagai outlier**.
  * `OrderCount` (**13.09%**) dan `CouponUsed` (**11.70%**) memiliki proporsi outlier cukup tinggi.
```python
target_col = 'Churn'
alpha = 0.05
y = df[target_col]

hasil = []

for col in cat_cols:
    if col == target_col:
        continue

    x = df[col]
    data = pd.DataFrame({'fitur': x, 'target': y})

    if data['fitur'].nunique() < 2 or data['target'].nunique() < 2:
        continue

    # Chi-square & Cramér's V
    tabel = pd.crosstab(data['fitur'], data['target'])
    chi2, p, dof, expected = stats.chi2_contingency(tabel)
    n_total = tabel.values.sum()
    min_dim = min(tabel.shape) - 1
    cramers_v = np.sqrt(chi2 / (n_total * min_dim)) if min_dim > 0 else np.nan

    hasil.append({
        'Fitur': col,
        'p-value': p,
        "Chi-Square": chi2,
        "Cramers_V": cramers_v
    })

hasil_df = pd.DataFrame(hasil).sort_values('p-value')
display(hasil_df.round({'Chi-Square': 2, 'p-value': 5, 'Cramers_V': 3}))
```

```python
# Visualisasi fitur signifikan
signifikan = hasil_df[hasil_df['p-value'] <= alpha].sort_values('Cramers_V', ascending=False)

if not signifikan.empty:
    plt.figure(figsize=(10, max(4, len(signifikan) * 0.4)))
    plt.barh(signifikan['Fitur'], signifikan['Cramers_V'], color='steelblue')
    plt.axvline(x=0.10, color='orange', linestyle='--', label='Lemah (0.10)')
    plt.axvline(x=0.30, color='red', linestyle='--', label='Kuat (0.30)')
    plt.gca().invert_yaxis()
    plt.xlabel("Cramér's V (Kekuatan Hubungan)")
    plt.title("Fitur Signifikan terhadap Churn (p ≤ 0.05)")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

## **Section 7. Model Development**

### **7.1 Initialization**

```python
def OHE(drop=None):
    kw = {"handle_unknown": "ignore"}
    if drop is not None:
        kw["drop"] = drop
    kw["sparse_output" if version.parse(sklearn.__version__) >= version.parse("1.2") else "sparse"] = False
    return OneHotEncoder(**kw)
```

Fungsi `OHE()` adalah pembungkus untuk `OneHotEncoder` Scikit-learn yang menjaga kompatibilitas antar versi (`sparse_output` vs `sparse`).

* `handle_unknown='ignore'` mencegah error saat ada kategori baru.
* `drop` opsional untuk menghindari *dummy variable trap*.
* Secara otomatis menyesuaikan parameter sesuai versi Scikit-learn yang digunakan.

```python
class NoOutlier(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None): return self
    def transform(self, X):    return X

class IQRClipper(BaseEstimator, TransformerMixin):
    def __init__(self, factor=1.5): self.factor=factor
    def fit(self, X, y=None):
        X=np.asarray(X, float); q1=np.nanpercentile(X,25,0); q3=np.nanpercentile(X,75,0); iqr=q3-q1
        self.lower_=q1-self.factor*iqr; self.upper_=q3+self.factor*iqr;
        return self
    def transform(self, X):
        return np.clip(np.asarray(X,float), self.lower_, self.upper_)

class QuantileWinsorizer(BaseEstimator, TransformerMixin):
    def __init__(self, lower=0.01, upper=0.99): self.lower, self.upper = lower, upper
    def fit(self, X, y=None):
        X=np.asarray(X,float); self.low_=np.nanpercentile(X, self.lower*100, 0); self.high_=np.nanpercentile(X, self.upper*100, 0)
        return self
    def transform(self, X):
        return np.clip(np.asarray(X,float), self.low_, self.high_)
```

Terdapat tiga pendekatan utama untuk menangani outlier:
- `NoOutlier`: Digunakan sebagai passthrough jika model tidak membutuhkan perlakuan terhadap outlier.
Hal ini menjaga fleksibilitas pipeline agar bisa menonaktifkan outlier treatment dengan mudah.
- `IQRClipper`: Metode clipping berbasis IQR (Interquartile Range).
Nilai di luar rentang [Q1−1.5×IQR, Q3+1.5×IQR] dipotong ke batas bawah/atas.
Cocok untuk model yang sensitif terhadap skala ekstrem seperti Logistic Regression atau KNN.
- `QuantileWinsorizer`: Alternatif Winsorization dengan batas persentil **(1%–99% secara default)**.
Cocok untuk data yang memiliki skewness tinggi atau distribusi non-normal,
misalnya durasi penggunaan aplikasi atau jumlah transaksi.

```python
class VIFSelector(BaseEstimator, TransformerMixin):
    """Drop kolom hingga seluruh VIF <= max_vif atau jumlah fitur mencapai min_features."""
    def __init__(self, max_vif=5.0, max_iter=100, min_features=5, verbose=False):
        self.max_vif=max_vif; self.max_iter=100 if max_iter is None else int(max_iter)
        self.min_features=int(min_features); self.verbose=verbose

    def _dense(self, X):
        import pandas as pd
        if isinstance(X, pd.DataFrame):
            return X.values.astype(float, copy=False), X.columns.to_list()
        X = X.toarray() if hasattr(X, "toarray") else np.asarray(X)
        return X.astype(float, copy=False), [f"f{i}" for i in range(X.shape[1])]

    def fit(self, X, y=None):
        X, cols = self._dense(X);
        keep = np.where(np.nanvar(X, axis=0) > 0)[0].tolist()
        it=0
        while it < self.max_iter and len(keep) > self.min_features:
            Xs = X[:, keep]
            try:
                vifs=[variance_inflation_factor(Xs, i) for i in range(Xs.shape[1])]
            except Exception:
                break
            vmax, imax = float(np.nanmax(vifs)), int(np.nanargmax(vifs))
            if self.verbose:
                print(f"[VIF] iter={it} max VIF={vmax:.3f} @ idx={imax}")
            if vmax <= self.max_vif:
                break
            del keep[imax]; it += 1
        self.keep_indices_ = np.array(keep, dtype=int)
        return self

    def transform(self, X):
        X,_ = self._dense(X)
        return X[:, self.keep_indices_]
```

`VIFSelector` digunakan untuk menghapus fitur numerik yang memiliki **multikolinearitas tinggi** berdasarkan **Variance Inflation Factor (VIF)**.

* Fitur dihapus secara iteratif hingga seluruh VIF ≤ `max_vif` (default: 5.0).
* `min_features` menjaga agar jumlah fitur tidak berkurang terlalu banyak.
* Cocok digunakan sebagai **feature selector** sebelum model berbasis regresi (misalnya Logistic Regression).
* `verbose=True` menampilkan log proses iterasi dan nilai VIF maksimum pada setiap langkah.

Semua helper ini nantinya digunakan di dalam pipeline seperti berikut:

* **OHE** → untuk categorical encoding.
* **Outlier Clipper / Winsorizer** → untuk stabilisasi numerik.
* **VIFSelector** → untuk pengurangan fitur korelatif.

Hasilnya adalah *pipeline* yang bersih, modular, dan konsisten di setiap eksperimen model, baik saat validasi silang maupun saat inferensi produksi.

```python
# Eksperimen imputasi
def preview_imputation_shapes(X, y, test_size=0.2, random_state=RANDOM_STATE):
    strategies = {
        "Iterative": ColumnTransformer(
            [("num", Pipeline([("imp", IterativeImputer(initial_strategy="median", random_state=random_state))]), num_cols),
             ("cat", OHE(drop='first'), cat_cols)], remainder="drop"
        ),
        "SimpleMedian": ColumnTransformer(
            [("num", Pipeline([("imp", SimpleImputer(strategy="median"))]), num_cols),
             ("cat", OHE(drop='first'), cat_cols)], remainder="drop"
        ),
        "KNNImputer": ColumnTransformer(
            [("num", Pipeline([("imp", KNNImputer())]), num_cols),
             ("cat", OHE(drop='first'), cat_cols)], remainder="drop"
        ),
        "DropNA": None  # akan by-pass transformer; kita drop NA di df lalu OHE saja
    }

    for name, ct in strategies.items():
        if name == "DropNA":
            df_drop = pd.concat([X, y], axis=1).dropna()
            Xd = df_drop.drop(columns=['Churn']); yd = df_drop['Churn']
            X_tr, X_te, y_tr, y_te = train_test_split(Xd, yd, test_size=test_size, random_state=random_state, stratify=yd)
            ohe = ColumnTransformer([("cat", OHE(drop='first'), cat_cols)], remainder="passthrough")
            Xt = ohe.fit_transform(X_tr)
            print(f"[{name}] X_train:{X_tr.shape}, X_test:{X_te.shape}, y_train:{y_tr.shape}, y_test:{y_te.shape} | features after OHE:{Xt.shape[1]}")
        else:
            X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=test_size, random_state=random_state, stratify=y)
            Xt = ct.fit_transform(X_tr)
            print(f"[{name}] X_train:{X_tr.shape}, X_test:{X_te.shape}, y_train:{y_tr.shape}, y_test:{y_te.shape} | features after prep:{Xt.shape[1]}")

preview_imputation_shapes(X, y)
```

* Ketiga metode imputasi (`Iterative`, `SimpleMedian`, `KNN`) menghasilkan jumlah observasi yang **sama** karena seluruh missing values berhasil diisi, tanpa kehilangan baris data.
* Jumlah fitur setelah *One-Hot Encoding* tetap **25 kolom** di semua strategi, artinya transformasi kategorikal tetap konsisten di seluruh eksperimen.
* Metode **DropNA** mengurangi ukuran data pelatihan menjadi sekitar **33% lebih sedikit** dari dataset awal, menunjukkan bahwa jumlah data dengan missing cukup signifikan. Hal ini kita jadikan pertimbangan untuk memilih strategi imputasi yang lebih bijak selain **DropNA**.

```python
class DropNaRows(BaseEstimator, TransformerMixin):
    def __init__(self, cols=None): self.cols = cols
    def fit(self, X, y=None): return self
    def transform(self, X):
        X_ = pd.DataFrame(X).copy()
        if self.cols is None:
            X_ = X_.dropna()
        else:
            X_ = X_.dropna(subset=self.cols)
        return X_.values
```

```python
# Loop eksperimen: Imputer × Outlier × Model (LR & RF)
def make_preprocessor(imputer: str, outlier: str, with_dropna=False):
    """
    imputer : 'iterative' | 'simple' | 'knn'
    outlier : 'none' | 'iqr' | 'winsor'
    with_dropna : if True, drop NA inside pipeline (per-fold) BEFORE imputation (generally keep False)
    """
    if imputer == "iterative":
        imp = IterativeImputer(initial_strategy="median", random_state=RANDOM_STATE)
    elif imputer == "simple":
        imp = SimpleImputer(strategy="median")
    elif imputer == "knn":
        imp = KNNImputer()
    else:
        raise ValueError("Unknown imputer")

    out_tr = NoOutlier() if outlier=="none" else IQRClipper() if outlier=="iqr" else QuantileWinsorizer()

    # numeric pipeline
    num_steps = []
    if with_dropna:
        num_steps.append(("dropna", DropNaRows()))
    num_steps += [
        ("impute", imp),
        ("outlier", out_tr),
        ("scale", RobustScaler())
    ]
    num_pipe = Pipeline(num_steps)

    # categorical pipeline
    cat_pipe = Pipeline([
        ("ohe", OHE(drop="first"))
    ])

    preproc = ColumnTransformer([
        ("num", num_pipe, num_cols),
        ("cat", cat_pipe, cat_cols)
    ], remainder="drop")

    return preproc

# Grid & model
imputer_grid = ["iterative", "simple", "knn"]
outlier_grid = ["none", "iqr", "winsor"]

base_model = LogisticRegression(random_state=RANDOM_STATE, max_iter=1000)

SCORING = {"f2": F2_SCORER}
rows = []

for imp in imputer_grid:
    for out in outlier_grid:
        preproc = make_preprocessor(imp, out, with_dropna=False)

        _ = preproc.fit(X_train, y_train)
        feat_count = preproc.transform(X_train).shape[1]

        pipe = Pipeline([
            ("prep", preproc),
            ("model", base_model)
        ])

        scores = cross_validate(
            pipe, X_train, y_train,
            cv=CV, scoring=SCORING,
            return_train_score=True, n_jobs=-1, error_score="raise"
        )

        rows.append({
            "Imputer": imp,
            "Outlier": out,
            "Features_after_prep": feat_count,

            # F2 Score
            "CV Train F2 (Mean)": scores["train_f2"].mean(),
            "CV Val F2 (Mean)":   scores["test_f2"].mean(),
            "CV Val F2 (Std)":    scores["test_f2"].std(),
        })

        print(f"[OK] imp={imp:9s} | out={out:6s} | features:{feat_count}")

results_cv = (
    pd.DataFrame(rows)
    .sort_values(["CV Val F2 (Mean)", "CV Val F2 (Std)"], ascending=[False, True])
    .reset_index(drop=True)
)

results_cv.style.highlight_max(
    subset="CV Val F2 (Mean)",
    color="darkred"
).highlight_min(
    subset="CV Val F2 (Std)",
    color="darkgreen"
).format({
    "CV Train F2 (Mean)": "{:.4f}",
    "CV Val F2 (Mean)":   "{:.4f}",
    "CV Val F2 (Std)":    "{:.4f}",
})
```

Eksperimen ini menguji pengaruh **strategi imputasi** dan **penanganan outlier** terhadap performa model menggunakan metrik **F2-score** pada skema **cross-validation**.
Tujuan utamanya adalah mencari kombinasi preprocessing yang memberikan keseimbangan antara **performansi (F2 Mean)** dan **stabilitas (F2 Std)**.

* Secara umum, **perbedaan antar strategi relatif kecil**, menunjukkan bahwa pipeline preprocessing sudah cukup robust terhadap variasi metode imputasi dan outlier.

Meskipun KNN tanpa outlier menghasilkan F2 tertinggi secara rata-rata, **Iterative + Winsorizer** lebih disarankan karena memberikan keseimbangan antara performa dan kestabilan model.
Dengan kata lain, kombinasi ini **lebih dapat diandalkan pada data baru**, terutama untuk kasus bisnis yang menuntut konsistensi hasil prediksi.

### **7.2 Developing the Model**

Setiap jenis model memiliki asumsi dan sensitivitas berbeda terhadap **skala, distribusi, dan korelasi fitur**.
Karena itu, setiap pipeline (P1–P3) dibangun dengan *treatment* yang disesuaikan dengan kebutuhan matematis modelnya.

#### **7.2.1 P1 - Linear & Distance-Based Models (Logistic Regression, KNN)**

*Treatment:* **Imputation → Winsorization → Scaling → Encoding + VIF Selection**

```python
# Build 3 preprocessors (P1/P2/P3)
clipper = QuantileWinsorizer()

# P1: LR/KNN: imputasi + winsorizer + SCALING + OHE(drop='first')
num_lr = Pipeline([
    ("impute", IterativeImputer(initial_strategy="median", random_state=RANDOM_STATE)),
    ("clip",   clipper),
    ("scale",  RobustScaler())
])
cat_lr = Pipeline([
    ("ohe", OHE(drop="first"))
])
prep_lr_knn = ColumnTransformer([
    ("num", num_lr,  num_cols),
    ("cat", cat_lr,  cat_cols)
], remainder="drop")
```

Model seperti Logistic Regression dan KNN **sangat sensitif terhadap nilai ekstrem** karena:
  * Regresi linear menggunakan [*ordinary least squares (OLS)*](https://arxiv.org/pdf/1404.6274), yang membesar secara kuadrat terhadap outlier.
  * KNN menghitung jarak Euclidean, yang berarti [*satu nilai ekstrem dapat mendominasi jarak antar titik*](https://www.datasciencebase.com/supervised-ml/algorithms/k-nearest-neighbors/common-mistakes/).
* Winsorization membatasi nilai ke persentil tertentu tanpa menghapus observasi, menjaga kestabilan varian dan mean.
* **RobustScaler** dipilih karena menskalakan data berdasarkan median dan IQR, bukan mean dan std, sehingga tidak terpengaruh outlier.

#### **7.2.2 P2 - Classical Tree-Based Models (Decision Tree, Random Forest, Gradient Boosting, AdaBoost)**

*Treatment:* **Imputation → Encoding (tanpa scaling atau outlier clipping)**

```python
# P2: Klasik Tree (DT/RF/GB/AB): imputasi + OHE, TANPA scaling
num_tree = Pipeline([
    ("impute", IterativeImputer(initial_strategy="median", random_state=RANDOM_STATE))
])
cat_tree = Pipeline([
    ("ohe", OHE(drop=None))       # drop=None aman untuk pohon
])
prep_tree = ColumnTransformer([
    ("num", num_tree, num_cols),
    ("cat", cat_tree, cat_cols)
], remainder="drop")
```

[*Mengapa Tidak Perlu Scaling*](https://stackoverflow.com/questions/29842647/feature-scaling-required-or-not):

* *Decision tree models* membuat pemisahan data berdasarkan **urutan nilai fitur**, bukan jarak antar titik seperti pada KNN atau regresi.
* Karena itu, perubahan skala (misalnya normalisasi atau standardisasi) **tidak mengubah urutan nilai**, sehingga **batas pemisahan (split)** pada pohon tetap sama sebelum dan sesudah scaling.
* Dengan kata lain, scaling tidak memberikan manfaat tambahan & justru hanya menambah langkah pemrosesan yang tidak perlu.

[*Mengapa Tidak Perlu Outlier Clipping*](https://communities.sas.com/t5/SAS-Communities-Library/Outliers-and-Tree-Based-Models-Should-We-Be-Concerned/ta-p/955438):

* Pada *decision tree model*, **outlier biasanya hanya masuk ke satu cabang tertentu**, sehingga pengaruhnya tidak menyebar ke seluruh struktur pohon.
* Pada model ensemble seperti Random Forest atau Gradient Boosting, efek outlier **teredam secara alami** karena setiap pohon dilatih pada subset data yang berbeda (*bagging*) dan hasil akhirnya merupakan **rata-rata dari banyak pohon** (*averaging*).
* Akibatnya, outlier jarang memberikan dampak besar terhadap performa keseluruhan model.

#### **7.2.3 P3 - Gradient Boosters (XGBoost, LightGBM, CatBoost)**

*Treatment:* **Imputation + Optional Scaling + OHE (drop =None)**

```python
# P3: Boosters (XGB/LGBM/CB): numeric passthrough (NaN), cat OHE
num_boost = Pipeline([
    ("impute", IterativeImputer(initial_strategy="median", random_state=RANDOM_STATE)),
    ("scale",  RobustScaler())
])
cat_boost = Pipeline([
    ("ohe", OHE(drop=None))
])
prep_boost = ColumnTransformer([
    ("num", num_boost, num_cols),
    ("cat", cat_boost, cat_cols)
], remainder="drop")
```

[*Mengapa Scaling Bersifat Opsional*](https://www.datasciencebase.com/supervised-ml/algorithms/gradient-boosting/XGBoost/common-mistakes/):

* Meski masih berbasis pohon, boosting melakukan **optimisasi numerik terhadap gradient loss**,
  sehingga perbedaan rentang nilai antar fitur dapat memengaruhi **kecepatan konvergensi**.
* Scaling moderat seperti *RobustScaler* dapat membantu stabilitas numerik,
  terutama jika fitur memiliki rentang yang sangat besar.

VIF Selection dilakukan karena:

  * Interpretasi feature importance menjadi **bias**.
* Seleksi fitur berbasis *Variance Inflation Factor (VIF)* menjaga [*stabilitas koefisien*](https://metricgate.com/blogs/vif-multicollinearity-regression/) tanpa mengurangi jumlah fitur terlalu banyak.

* **LightGBM** dapat menjadi alternatif ringan dengan performa yang hampir setara dan waktu pelatihan yang lebih efisien.

```python
resampling_options = {
    "None":                None,  # untuk class-weighted (tanpa resampling)
    "RandomUnderSampler":  RandomUnderSampler(random_state=RANDOM_STATE),
    "RandomOverSampler":   RandomOverSampler(random_state=RANDOM_STATE),
    "SMOTE":               SMOTE(random_state=RANDOM_STATE),
}

# hitung class weight dari TRAIN
classes = np.array([0, 1])
cw_vals = compute_class_weight(class_weight='balanced', classes=classes, y=y_train)
cw_map  = {int(c): float(w) for c, w in zip(classes, cw_vals)}

n_pos   = int((y_train == 1).sum())
n_neg   = int((y_train == 0).sum())
xgb_spw = n_neg / max(n_pos, 1)

print(f"class_weight LGBM: {cw_map} | XGB scale_pos_weight: {xgb_spw:.3f}")
```

```python
model_options = {
    "XGB_plain":   XGBClassifier(random_state=RANDOM_STATE, n_estimators=300),
    "XGB_cw":      XGBClassifier(random_state=RANDOM_STATE, n_estimators=300, scale_pos_weight=xgb_spw),
    "LGB_plain":   lgb.LGBMClassifier(random_state=RANDOM_STATE, n_estimators=300),
    "LGB_cw":      lgb.LGBMClassifier(random_state=RANDOM_STATE, n_estimators=300, class_weight=cw_map),
}

SCORING = {"f2": F2_SCORER}
rows = []

for r_name, r_step in resampling_options.items():
    for m_name, base_model in model_options.items():
        if r_name != "None" and m_name.endswith("_cw"):
            continue # SKIP kombinasi resampling + class_weight model (double balancing)

        print(f"Resampling: {r_name:<15} | Model: {m_name:<10}")

        pipe = ImbPipeline([
            ("prep",       clone(prep_boost)),
            ("resampling", clone(r_step) if r_step else 'passthrough'),
            ("model",      clone(base_model))
        ])

        scores = cross_validate(
            pipe, X_train, y_train,
            cv=CV, scoring=SCORING,
            return_train_score=True, n_jobs=-1, error_score="raise"
        )

        rows.append({
            "Model": m_name,
            "Resampling": r_name,
            "CV Train F2 (Mean)": scores["train_f2"].mean(),
            "CV Val F2 (Mean)":   scores["test_f2"].mean(),
            "CV Val F2 (Std)":    scores["test_f2"].std(),
        })

results_all = (
    pd.DataFrame(rows)
      .sort_values(["CV Val F2 (Mean)", "CV Val F2 (Std)"], ascending=[False, True])
      .reset_index(drop=True)
)

results_all.style.highlight_max(
    "CV Val F2 (Mean)",
    color="darkred").format({
        "CV Train F2 (Mean)": "{:.4f}",
        "CV Val F2 (Mean)":   "{:.4f}",
        "CV Val F2 (Std)":    "{:.4f}",
    })
```

```python
best = results_all.iloc[0]
best_pipe_lgb_xgb = ImbPipeline([
    ("prep",       clone(prep_boost)),
    ("resampling", clone(resampling_options[best["Resampling"]]) if best["Resampling"]!="None" else 'passthrough'),
    ("model",      clone(model_options[best["Model"]]))
])
best_pipe_lgb_xgb.fit(X_train, y_train)

y_pred = best_pipe_lgb_xgb.predict(X_test)
print("FINAL TEST F2:", fbeta_score(y_test, y_pred, beta=2))
```

ROS/SMOTE memang menambah *recall* pada sebagian fold, tetapi **tidak melampaui kombinasi class weight** dan membawa risiko *shift* dari distribusi asli. Karena tujuan utamanya **recall tinggi yang stabil**, pilihan **class weight** adalah yang paling konsisten dan kuat untuk deployment.

### **7.3 Hyperparameter Tuning**

```python
xgb_base = Pipeline([
    ('transformer', clone(prep_boost)),
    ('model', XGBClassifier(
        random_state=RANDOM_STATE,
        scale_pos_weight=xgb_spw,
    ))
])

lgbm_base = Pipeline([
    ('transformer', clone(prep_boost)),
    ('model', lgb.LGBMClassifier(
        random_state=RANDOM_STATE,
        class_weight=cw_map,
    ))
])
```

```python
xgb_params = {
    "model__n_estimators":     [50, 100, 150],
    "model__max_depth":        [3, 5, 7, 9],
    "model__learning_rate":    [0.01, 0.05, 0.1, 0.2],
    "model__subsample":        [0.8, 0.9, 1.0],
    "model__colsample_bytree": [0.8, 0.9, 1.0],
    "model__min_child_weight": [1, 3, 5]
}

lgb_params = {
    "model__n_estimators":      [50, 100, 150],
    "model__learning_rate":     [0.02, 0.05, 0.10],
    "model__num_leaves":        [31, 63, 95],
    "model__min_child_samples": [20, 40, 80],
    "model__subsample":         [0.8, 1.0],
    "model__colsample_bytree":  [0.8, 1.0]
}
```

```python
inner = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
outer = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

def nested_cv_summary(model_name, base_estimator, param_dist, inner_cv, outer_cv, n_iter=88):
    """
    Nested CV:
      - Outer loop: unbiased performance estimate (mean ± std)
      - Inner loop: hyperparameter search (RandomizedSearchCV)
    Tidak melakukan prediksi pada test set sama sekali.
    Mengembalikan: (df_row, best_estimator_fitted_on_full_train)
    """
    search = RandomizedSearchCV(
        estimator=clone(base_estimator),
        param_distributions=param_dist,
        n_iter=n_iter,
        scoring=F2_SCORER,
        cv=inner_cv,
        n_jobs=-1,
        random_state=RANDOM_STATE,
        refit=True,
        verbose=1,
        return_train_score=False,
        error_score="raise"
    )

    outer_scores = cross_val_score(
        search, X_train, y_train,
        cv=outer_cv,
        scoring=F2_SCORER,
        n_jobs=-1
    )

    # Fit sekali (tanpa sentuh test)
    search.fit(X_train, y_train)
    best_est = search.best_estimator_
    best_params = search.best_params_

    row = {
        "Model":                    model_name,
        "Nested CV F2 (Mean)":      float(np.mean(outer_scores)),
        "Nested CV F2 (Std)":       float(np.std(outer_scores, ddof=0)),
        "Refit Best Params":        best_params
    }
    return pd.DataFrame([row]), best_est
```

```python
# Tuning XGB & LGBM
res_xgb, best_xgb = nested_cv_summary("XGBoost (class-weight)",  xgb_base,  xgb_params,  inner, outer, n_iter=88)
res_lgb, best_lgb = nested_cv_summary("LightGBM (class-weight)", lgbm_base, lgb_params, inner, outer, n_iter=88)
```

| **Nested CV F2 (Std)**  | **0.0194** | Konsistensi performa antar-fold. Nilai yang rendah menandakan model relatif stabil terhadap variasi sampel pelatihan/validasi.                                               |

Berdasarkan hasil tersebut, **XGBoost (class-weight)** menjadi kandidat utama untuk deployment karena:

### **7.4 Threshold Tuning (PR Curve)**

```python
def pr_curve_cv_plot(pipe, X, y, *, beta=2.0, cv=None, label=None, random_state=5):
    if cv is None:
        cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=random_state)

    precisions, recalls, aps, best_thr = [], [], [], []

    for tr_idx, va_idx in cv.split(X, y):
        X_tr, X_va = X.iloc[tr_idx], X.iloc[va_idx]
        y_tr, y_va = y.iloc[tr_idx], y.iloc[va_idx]

        model = clone(pipe).fit(X_tr, y_tr)
        if hasattr(model, "predict_proba"):
            y_score = model.predict_proba(X_va)[:, 1]
        elif hasattr(model, "decision_function"):
            y_score = model.decision_function(X_va)
        else:
            raise ValueError("Estimator tidak memiliki predict_proba/decision_function")

        precision, recall, thr = precision_recall_curve(y_va, y_score)
        ap = average_precision_score(y_va, y_score)

        beta2 = beta**2
        denom = (beta2 * precision[:-1] + recall[:-1] + 1e-12)
        fbeta_vals = (1 + beta2) * precision[:-1] * recall[:-1] / denom
        best_idx = int(np.nanargmax(fbeta_vals))
        thr_star = float(thr[best_idx]) if len(thr) > 0 else 0.5

        precisions.append(np.interp(np.linspace(0, 1, 100), recall[::-1], precision[::-1]))
        recalls.append(np.linspace(0, 1, 100))
        aps.append(ap)
        best_thr.append(thr_star)

    precisions = np.vstack(precisions)
    mean_precision = precisions.mean(axis=0)
    std_precision = precisions.std(axis=0)
    mean_recall = recalls[0]
    mean_ap = np.mean(aps)
    median_thr = np.median(best_thr)
    baseline = np.mean(y)

    fig, ax = plt.subplots(figsize=(7, 6))
    ax.plot(mean_recall, mean_precision, lw=2,
            label=f"{label or 'Model'} | Mean AUC-PR = {mean_ap:.3f}")
    ax.fill_between(mean_recall,
                    np.maximum(mean_precision - std_precision, 0),
                    np.minimum(mean_precision + std_precision, 1),
                    alpha=0.2, label="±1 std")
    ax.hlines(baseline, 0, 1, linestyles="--", color="gray", label=f"Baseline={baseline:.3f}")
    ax.set_xlim(0, 1); ax.set_ylim(0, 1.05)
    ax.set_xlabel("Recall"); ax.set_ylabel("Precision")
    ax.set_title(f"Cross-Validated Precision–Recall Curve (Train-only)\nβ={beta}, Median Thr={median_thr:.3f}")
    ax.grid(True); ax.legend(loc="lower left")
    plt.tight_layout(); plt.show()

    return {
        "Model": label or "Model",
        "AUC_PR (Mean)": round(mean_ap, 3),
        "AUC_PR (Std)": round(np.std(aps, ddof=0), 3),
        f"F{int(beta)} Thr (Median)": round(median_thr, 3),
        "Baseline": round(baseline, 3)
    }
```

```python
final_pipe = clone(best_xgb).fit(X_train, y_train)

cv_result = pr_curve_cv_plot(
    final_pipe, X_train, y_train,
    beta=2.0, label="XGBoost (class-weight)", random_state=RANDOM_STATE
)
best_threshold = cv_result["F2 Thr (Median)"]

print(f"OK. Final model = XGBoost. Median F2-threshold = {best_threshold:.3f}")

display(pd.DataFrame([cv_result]))
```

Dari hasil di atas dapat dilihat bahwa nilai **AP** atau **Average Precision Score (AUC–PR)** adalah **0.945**.
Artinya, model **sangat efisien** dalam mengidentifikasi pelanggan yang benar-benar berpotensi churn, jauh di atas baseline prevalensi positif sebesar **0.168**.

Nilai **F₂ = 0.986** pada **threshold = 0.315** menunjukkan bahwa model mencapai **keseimbangan optimal antara recall dan precision** ketika probabilitas churn melebihi 31,5%. Dengan ambang ini, model mampu menangkap sebagian besar pelanggan yang akan churn tanpa terlalu banyak menghasilkan prediksi positif palsu.

Meskipun nilai **F₂ maksimum** dari kurva Precision–Recall menggambarkan potensi performa tertinggi model pada ambang optimal, **evaluasi utama tetap mengacu pada metrik benchmark hasil cross-validation** (Nested CV F₂). Benchmark tersebut merepresentasikan performa yang lebih stabil, konsisten antar-fold, dan realistis terhadap kondisi operasional sebenarnya.
PR Curve sendiri digunakan sebagai **analisis diagnostik** untuk memahami trade-off antara recall dan precision serta membantu penetapan threshold implementasi yang lebih presisi.
Dengan demikian, benchmark F₂ dari proses validasi tetap dipertahankan sebagai metrik utama untuk menjaga **obyektivitas dan reproducibility** hasil evaluasi model.

### **8.1 Final Model**


```python
# Build Preprocessor (P3)
def build_prep_boost(num_cols: List[str], cat_cols: List[str], random_state: int = 5) -> ColumnTransformer:
    num_pipe = Pipeline([
        ("impute", IterativeImputer(initial_strategy="median", random_state=random_state)),
        ("scale",  RobustScaler())
    ])
    cat_pipe = Pipeline([
        ("ohe", OHE(drop=None))
    ])
    return ColumnTransformer([
        ("num", num_pipe, num_cols),
        ("cat", cat_pipe, cat_cols)
    ], remainder="drop")

# FINAL MODEL
@dataclass
class ChurnXGBCWConfig:
    random_state: int = 5
    threshold: float = 0.50
    best_params: Optional[Dict] = None
    num_cols: Optional[List[str]] = None
    cat_cols: Optional[List[str]] = None
    scale_pos_weight: Optional[float] = None  # if None, will be computed from y_train


class ChurnXGBCWPipeline:
    """
    Final pipeline for deployment (class-weight balancing):
      - Preprocessing (IterativeImputer + RobustScaler + OHE)
      - XGBClassifier with scale_pos_weight (no resampling)
      - Save/load using pickle (.sav)
      - Optional: export processed design matrix (no resampling)
    """
    def __init__(self, cfg: ChurnXGBCWConfig):
        assert cfg.num_cols is not None and cfg.cat_cols is not None, "num_cols & cat_cols must be defined"
        self.cfg = cfg
        self.prep_ = build_prep_boost(cfg.num_cols, cfg.cat_cols, random_state=cfg.random_state)

        # Model skeleton; scale_pos_weight will be set at fit-time if not provided
        self.model_ = XGBClassifier(
            random_state=cfg.random_state,
            n_jobs=-1
        )
        if cfg.best_params:
            clean_params = {
                k: (float(v) if isinstance(v, (np.floating,)) else int(v) if isinstance(v, (np.integer,)) else v)
                for k, v in cfg.best_params.items()
            }
            self.model_.set_params(**clean_params)

        self.pipe_ = Pipeline(steps=[
            ("transformer", self.prep_),
            ("model",       self.model_)
        ])

    def fit(self, X: pd.DataFrame, y: Union[pd.Series, np.ndarray]):
        """Train model (sets scale_pos_weight if not provided)."""
        self.X_train_ = X.copy()
        self.y_train_ = pd.Series(y).copy()

        # compute scale_pos_weight from TRAIN if not provided
        if self.cfg.scale_pos_weight is None:
            n_pos = int((self.y_train_ == 1).sum())
            n_neg = int((self.y_train_ == 0).sum())
            spw = n_neg / max(n_pos, 1)
            self.pipe_.named_steps["model"].set_params(scale_pos_weight=spw)
        else:
            self.pipe_.named_steps["model"].set_params(scale_pos_weight=self.cfg.scale_pos_weight)

        self.pipe_.fit(X, y)
        return self

    def predict_proba(self, X: pd.DataFrame) -> np.ndarray:
        return self.pipe_.predict_proba(X)

    def predict(self, X: pd.DataFrame) -> np.ndarray:
        proba = self.predict_proba(X)[:, 1]
        return (proba >= self.cfg.threshold).astype(int)

    def set_threshold(self, thr: float):
        self.cfg.threshold = float(thr)

    def export_processed(self, path: str):
        """
        Export preprocessed training design matrix (no resampling).
        Useful for inspection or feature importance analysis.
        """
        if not hasattr(self, "pipe_"):
            raise RuntimeError("Pipeline is not fitted yet.")
        print("[INFO] Transforming and exporting processed data...")
        transformer = self.pipe_.named_steps["transformer"]
        X_proc = transformer.transform(self.X_train_)
        if hasattr(X_proc, "toarray"):   # sparse?
            Xp = X_proc.toarray()
        else:
            Xp = np.asarray(X_proc)
        try:
            cols = transformer.get_feature_names_out()
        except Exception:
            cols = [f"f_{i}" for i in range(Xp.shape[1])]
        df_proc = pd.DataFrame(Xp, columns=cols)
        df_proc["Churn"] = self.y_train_.values
        os.makedirs(os.path.dirname(path), exist_ok=True)
        df_proc.to_csv(path, index=False)
        print(f"[SAVED] Processed data exported to: {path}")

    def save(self, path: str):
        """Save trained pipeline to .sav using pickle."""
        os.makedirs(os.path.dirname(path), exist_ok=True)
        payload = {
            "cfg": self.cfg,
            "pipe": self.pipe_,
            "threshold": float(getattr(self.cfg, "threshold", 0.5))
        }
        with open(path, "wb") as f:
            pickle.dump(payload, f)
        print(f"[SAVED] Model saved to: {os.path.abspath(path)}")

    @classmethod
    def load(cls, path: str) -> "ChurnXGBCWPipeline":
        with open(path, "rb") as f:
            obj = pickle.load(f)
        inst = cls(obj["cfg"])
        inst.pipe_ = obj["pipe"]
        thr = obj.get("threshold", getattr(inst.cfg, "threshold", 0.5))
        inst.cfg.threshold = float(thr)
        print(f"[LOADED] Model loaded from: {path} | threshold={inst.cfg.threshold:.3f}")
        return inst
```

### **9.2 Model Limitations**

Beberapa hal yang mungkin menjadi batasan daripada model adalah sebagai berikut :

* **Definisi label churn yang tidak sepenuhnya pasti :**
  Penetapan label churn didasarkan pada asumsi periode inaktivitas pelanggan tanpa adanya definisi bisnis yang benar-benar eksplisit. Hal ini menimbulkan potensi **bias label**, di mana pelanggan yang masih aktif namun belum melakukan transaksi dalam jangka waktu tertentu dapat salah diklasifikasikan sebagai churn. Akibatnya, akurasi model terhadap kondisi aktual pelanggan bisa berkurang.

* **Fokus penggunaan metrik yang menekankan pada recall (F₂-Score) :**
  Model dioptimalkan menggunakan **F₂-Score** yang menitikberatkan pada kemampuan mendeteksi seluruh pelanggan churn (*recall*). Pendekatan ini membuat model menjadi **kurang sesuai apabila fokus bisnis bergeser ke precision**, misalnya ketika perusahaan ingin meminimalkan biaya retensi yang salah sasaran pada pelanggan non-churn.

* **Tidak mempertimbangkan adanya pengaruh musiman (seasonality) :**
  Model belum memperhitungkan perilaku pelanggan yang dipengaruhi oleh faktor musiman, seperti promosi tertentu, periode liburan, atau tren tahunan, karena **tidak adanya fitur temporal (time-based feature)** dalam data pelatihan. Hal ini dapat menyebabkan model kesulitan beradaptasi terhadap perubahan pola transaksi musiman.

* **Kurangnya fitur aspek finansial (monetary features) :**
  Model belum mencakup fitur-fitur yang merepresentasikan **nilai ekonomi pelanggan**, seperti total pengeluaran, rata-rata nilai transaksi, atau *customer lifetime value (CLV)*. Ketiadaan aspek ini membuat strategi retensi berpotensi **tidak efisien**, karena anggaran bisa dialokasikan untuk pelanggan bernilai rendah yang dampak retensinya tidak signifikan terhadap profitabilitas perusahaan.

* **Retraining berkala**: **triwulan** atau saat drift terdeteksi; audit ulang metrik **F2/AP** dan FPR.
* **Insight lanjutan**: gunakan **SHAP interaction** & **partial dependence** untuk pasangan kunci (mis. *Complain × Low Tenure*).

#### **10.3.2 For CMO (Chief Marketing Officer)**

   Segmentasi **merah–kuning–hijau** berbasis skor; jaga **false alarm (~1.6%)**. Terapkan **monthly spend cap** pada setiap segmen.
